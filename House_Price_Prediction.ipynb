```python
none_cols = [
    "PoolQC", "MiscFeature", "Alley", "Fence",
    "MasVnrType", "BsmtQual", "BsmtCond",
    "BsmtExposure", "BsmtFinType1", "BsmtFinType2",
    "FireplaceQu", "GarageType", "GarageFinish",
    "GarageQual", "GarageCond"
]

for col in none_cols:
    df[col] = df[col].fillna("None")

df["MasVnrArea"] = df["MasVnrArea"].fillna(0)

df["LotFrontage"] = df["LotFrontage"].fillna(df["LotFrontage"].median())

df["Electrical"] = df["Electrical"].fillna(df["Electrical"].mode()[0])
```

```python
df["HasGarage"] = df["GarageYrBlt"].notna().astype(int)

df["GarageYrBlt"] = df["GarageYrBlt"].fillna(
    df["GarageYrBlt"].median()
)
```

```python
df.head()
```

```python
x = df.drop("SalePrice", axis = 1)
y = df["SalePrice"]
```

```python
print(x.shape)
print(y.shape)
```

```python
categorical_columns = x.select_dtypes(include=["object"]).columns
numerical_columns = x.select_dtypes(exclude=["object"]).columns

print("Categorical:",len(categorical_columns))
print("Numerical:",len(numerical_columns))
```

```python
print(categorical_columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)
```

```python
x_train_processed = preprocessor.fit_transform(x_train)
x_test_processed = preprocessor.transform(x_test)
```

```python
print("Before:", x_train.shape)
print("After:", x_train_processed.shape)
print("Test:", x_test_processed.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(x_train_processed,y_train)
```

```python
y_pred = model.predict(x_test_processed)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
import numpy as np
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse =  np.sqrt(mean_squared_error(y_test,y_pred))
r2 = r2_score(y_test,y_pred)

print("MAE:",mae)
print("MSE:",mse)
print("RMSE:",rmse)
print("R2:",r2)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Sale Price")
plt.ylabel("Predicted Sale Price")
plt.title("Actual vs Predicted SalePrice")

plt.show()
```
